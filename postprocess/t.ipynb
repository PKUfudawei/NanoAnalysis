```python
import os
import matplotlib
import shutil
from pathlib import Path

# 设置 PATH
os.environ['PATH'] = '/usr/local/bin:' + os.environ.get('PATH', '')
os.environ['PATH'] = os.environ['PATH'] + ':' + os.path.expanduser('~/Library/TinyTeX/bin/universal-darwin')
```

```python

# 清理缓存
tex_cache = Path(matplotlib.get_cachedir()) / 'tex.cache'
if tex_cache.exists():
    shutil.rmtree(tex_cache)
    print(f"已删除缓存: {tex_cache}")

# 测试 LaTeX 渲染
import matplotlib.pyplot as plt
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
})

plt.figure(figsize=(4, 2))
plt.text(0.5, 0.5, r'$E = mc^2$', fontsize=20, ha='center', va='center')
plt.title('LaTeX Test')
plt.show()
print("✅ 成功！")
```
